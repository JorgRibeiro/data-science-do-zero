### Programação Orientada a Objetos

```python
class CountigClicker:
    """A classe pode/deve ter um docstring, como as funções"""

    def __init__(self, count = 0):
        self.count = count

    def __repr__(self):
        return f"CountigClicker(count={self.count})"
    
    def click(self, num_times = 1):
        """Clique no contador algumas vezes."""
        self.count += num_times

    def read(self):
        return self.count
    
    def reset(self):
        self.count = 0
```

```python
clicker1 = CountigClicker()                 # inicializando em 0
clicker2 = CountigClicker(100)              # começa em count=100
clicker3 = CountigClicker(count=100)        # forma mais expressa de fazer a mesma coisa
```

```python
clicker = CountigClicker()
assert clicker.read() == 0
clicker.click()
clicker.click()
assert clicker.read() == 2
clicker.reset()
assert clicker.read() == 0
```

```python
# A subclasse herda rodo o comportamento da classe pai
class NoResetClicker(CountigClicker):
    # Esta classe tem os mesmo métodos da CountigClicker

    def reset(self):
        pass

clicker2 = NoResetClicker()
assert clicker2.read() == 0
clicker2.click()
assert clicker2.read() == 1
clicker2.reset()
assert clicker2.read() == 1
```
