### Funções 

```python
def double(x):
    """
    Nesse ponto, você coloca um docstring opcional para descrever a função.
    Por exemplo, essa função multiplica a entrada por 2.
    """
    return x * 2
```

```python
def apply_to_one(f):
    """Chama a função f usando 1 como argumento"""
    return f(1)

my_double = double              # refere-se a função X já definida
x = apply_to_one(my_double)

x
```

```python
y = apply_to_one(lambda x: x + 4)

y
```

```python
another_double = lambda x: 2 * 2        # não faça isso

def another_double(x):
    """Faça isso"""
    return x * 2
```
