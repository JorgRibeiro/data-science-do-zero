### Programação Orientada a Objetos

A orientação a objetos organiza dados e comportamentos em classes, facilitando modelar conceitos do domínio.


```python
class CountigClicker:
    """A classe pode/deve ter um docstring, como as funções"""

    def __init__(self, count = 0):
        self.count = count

    def __repr__(self):
        return f"CountigClicker(count={self.count})"
    
    def click(self, num_times = 1):
        """Clique no contador algumas vezes."""
        self.count += num_times

    def read(self):
        return self.count
    
    def reset(self):
        self.count = 0
```

```python
clicker1 = CountigClicker()                 # inicializando em 0
clicker2 = CountigClicker(100)              # começa em count=100
clicker3 = CountigClicker(count=100)        # forma mais expressa de fazer a mesma coisa
```

```python
clicker = CountigClicker()
assert clicker.read() == 0
clicker.click()
clicker.click()
assert clicker.read() == 2
clicker.reset()
assert clicker.read() == 0
```

```python
# A subclasse herda rodo o comportamento da classe pai
class NoResetClicker(CountigClicker):
    # Esta classe tem os mesmo métodos da CountigClicker

    def reset(self):
        pass

clicker2 = NoResetClicker()
assert clicker2.read() == 0
clicker2.click()
assert clicker2.read() == 1
clicker2.reset()
assert clicker2.read() == 1
```

### args e kwargs

args e kwargs permitem escrever funções que recebem quantidades variáveis de argumentos posicionais e nomeados.


```python
def doubler(f):
    # Aqui, definimos uma nova função que mantém uma referência a f
    def g(x):
        return 2 * f(x)
    # E retorna a nova função
    return g
```

```python
def f1(x):
    return 1 + x

g = doubler(f1)
assert g(3) == 8, "(3 + 1) * 2 should equal 8"
assert g(-1) == 0, "(-1 + 1) * 2 should equal 0"
```

```python
def f2(x, y):
    return x + y

g = doubler(f2)
try:
    g(1, 2)
except TypeError:
    print("as defined, g only takes one argument")
```

```python
def magic(*args, **kwargs):
    print("unnamed args:" , args)
    print("keyword args:" , kwargs)

magic(1, 2, key="word", key2="word2")
```

```python
def other_way_magic(x, y, z):
    return x + y + z

x_y_list = [1, 2]
z_dict = {"z": 3}

assert other_way_magic(*x_y_list, **z_dict) == 6, "1 + 2 + 3 should be 6"
```

```python
def doubler_correct(f):
    """funciona para qualquer entrada recebida por f"""
    def g(*args, **kwargs):
        """todo argumento fornecido para g deve ser transmitido para f"""
        return 2 * f(*args, **kwargs)
    
    return g

g = doubler_correct(f2)

assert g(1, 2) == 6, "doubler should work now"
```

```python
from typing import Callable

# A dica de tipo indica que o repetidor é uma função que recebe dois argumentos, uma string e um int, e retorna uma string.
def twice(repeater: Callable[[str, int], str], s: str) -> str:
    return repeater(s, 2)

def comma_reapeter(s: str, n: int) -> str:
    n_copies = [s for _ in range(n)]
    return ', '.join(n_copies)

assert twice(comma_reapeter, "type hints") == "type hints, type hints"
```
